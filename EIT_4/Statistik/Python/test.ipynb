```python
# Aufgabenstellung:
# Du hast drei verschiedene Glücksspielvarianten mit einem Einsatz von 10 €.
# Berechne für jede Variante den Erwartungswert analytisch und vergleiche ihn mit
# einer Monte-Carlo-Simulation.
#
# Spiel 1:
# - Einsatz: 10 €
# - 10 Münzwürfe
# - Bei Kopf: +3 €
# - Bei Zahl: -1 €
#
# Spiel 2:
# - Einsatz: 10 €
# - 10 Münzwürfe
# - Bei Kopf: aktuelles Guthaben wird mit 1.2 multipliziert
# - Bei Zahl: aktuelles Guthaben wird mit 0.8 multipliziert
#
# Spiel 3:
# - Einsatz: 10 €
# - Startgeld: 1 €
# - Es wird so lange geworfen, bis die erste Zahl erscheint.
# - Bei Kopf verdoppelt sich das Geld und das Spiel geht weiter.
# - Bei Zahl endet das Spiel und der derzeitige Betrag wird ausgezahlt.

import random

NUM_SIMULATIONS = 2_000_000


def simulieren_spiel1():
    profit = 0
    for _ in range(10):
        if random.random() < 0.5:
            profit += 3
        else:
            profit -= 1
    return 10 + profit


def erwartungswert_spiel1():
    durchschnitt_pro_wurf = 0.5 * 3 + 0.5 * (-1)
    return 10 + 10 * durchschnitt_pro_wurf


def simulieren_spiel2():
    betrag = 10.0
    for _ in range(10):
        if random.random() < 0.5:
            betrag *= 1.2
        else:
            betrag *= 0.8
    return betrag


def erwartungswert_spiel2():
    faktor_erwartung = 0.5 * 1.2 + 0.5 * 0.8
    return 10.0 * faktor_erwartung ** 10


def simulieren_spiel3():
    betrag = 1.0
    while True:
        if random.random() < 0.5:
            betrag *= 2
        else:
            return betrag


def erwartungswert_spiel3():
    return 1.0


def monte_carlo(sim_func, num_runs=NUM_SIMULATIONS):
    ergebnisse = [sim_func() for _ in range(num_runs)]
    return sum(ergebnisse) / len(ergebnisse)


print("=== Aufgabenstellung ===")
print("Spiel 1: Einsatz 10 €, 10 Münzwürfe, Kopf +3 €, Zahl -1 €")
print("Spiel 2: Einsatz 10 €, 10 Münzwürfe, Kopf x1.2, Zahl x0.8")
print("Spiel 3: Einsatz 10 €, Startgeld 1 €, Kopf verdoppelt, Spiel endet bei Zahl")
print()

print("=== Analytische Erwartungswerte ===")
print(f"Spiel 1: Erwarteter Endbetrag = {erwartungswert_spiel1():.2f} €, erwarteter Gewinn = {erwartungswert_spiel1() - 10:.2f} €")
print(f"Spiel 2: Erwarteter Endbetrag = {erwartungswert_spiel2():.2f} €, erwarteter Gewinn = {erwartungswert_spiel2() - 10:.2f} €")
print(f"Spiel 3: Erwarteter Endbetrag = {erwartungswert_spiel3():.2f} €, erwarteter Gewinn = {erwartungswert_spiel3() - 10:.2f} €")
print()

print("=== Monte-Carlo-Simulation (200.000 Durchläufe) ===")
print(f"Spiel 1: Simulierter Durchschnitt = {monte_carlo(simulieren_spiel1):.2f} €")
print(f"Spiel 2: Simulierter Durchschnitt = {monte_carlo(simulieren_spiel2):.2f} €")
print(f"Spiel 3: Simulierter Durchschnitt = {monte_carlo(simulieren_spiel3):.2f} €")
print()

print("Hinweis: Bei Spiel 3 ist der Einsatz 10 € die Teilnahmegebühr. Der erwartete Rückzahlungsbetrag beträgt 1 €.")
```
