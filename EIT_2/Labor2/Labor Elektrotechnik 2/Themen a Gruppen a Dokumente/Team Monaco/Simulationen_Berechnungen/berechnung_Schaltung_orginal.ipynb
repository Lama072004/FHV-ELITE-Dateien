```python
import numpy as np
```

```python
# Eingabe Werte BC337

U = 12              # Eingangsspannung DC
B = 400             # Muss anhand des Ic in der Grafik nachgesehen werden
Ic = 0.01           # Typisch für kleinverstärcker 5-20mA
FE = 0.05           # 0.05 - 0.1 Üblich, Standart 0.1, kleinere FE--> Höhere Spannungs verstärkung
m = 5               # 3 - 10 Üblich
Ube = 0.7           # Silizium = 0.7
```

```python
def format_engineering(value, unit=''):
    prefixes = [
        (1e9, 'G'),   # Giga
        (1e6, 'M'),   # Mega
        (1e3, 'k'),   # Kilo
        (1, ''),      # keine Vorsilbe
        (1e-3, 'm'),  # Milli
        (1e-6, 'µ'),  # Mikro
        (1e-9, 'n'),  # Nano
        (1e-12, 'p')  # Piko
    ]
    
    for factor, prefix in prefixes:
        if abs(value) >= factor or factor == 1e-12:
            formatted_value = value / factor
            return f"{formatted_value:.3f} {prefix}{unit}"
    return f"{value} {unit}"  # Falls gar nichts passt


Rc = U/(2*Ic)
Re = (FE*U)/Ic
Ue0 = Re*Ic
Ub0 = Ube + Ue0
Ib = Ic/B
Iq = m*Ib 
R2 = Ub0/Iq
R1 = (U-Ub0)/(Iq+Ib)
Uv = Rc/Re


print("Ic:              ",format_engineering(Ic, unit='A'))
print("Ib:              ",format_engineering(Ib, unit='A'))
print("Iq/IR2:          ",format_engineering(Iq, unit='A'))
print("Rc:              ",format_engineering(Rc, unit='Ω'))
print("Re:              ",format_engineering(Re, unit='Ω'))
print("R2:              ",format_engineering(R2, unit='Ω'))
print("R1:              ",format_engineering(R1, unit='Ω'))
print("U Verstäckung:   ", Uv)
```
