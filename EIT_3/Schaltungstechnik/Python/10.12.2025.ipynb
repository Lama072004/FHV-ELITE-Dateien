# Beispiel mit ElementarZweipolen

R-C-C-R Schaltung

```python
import numpy as np
import matplotlib as plt
pi = np.pi
j = 1j

R1 = 1000.
C1 = 22e-9
R2 = 100.
C2 = 22e-9
RV = 10e3


w = 1000*2*pi
ZC1 = 1/(j*w*C1)
ZC2 = 1/(j*w*C2)


A_R1 = np.array([[1,R1],[0,1]])
A_C1 = np.array([[1,0],[1/ZC1,1]])
A_C2 = np.array([[1,ZC2],[0,1]])
A_R2 = np.array([[1,0],[1/R2,1]])

A_ges = A_R1 @ A_C1 @ A_C2 @ A_R2

print("Gesamtmatrix A:")
print(A_ges)

A11 = A_ges[0,0]
A12 = A_ges[0,1]
A21 = A_ges[1,0]
A22 = A_ges[1,1]

Tu = 1/(A11 + 1/RV * A12)
print("Übertragungsfunktion Tu:")
print(Tu)
print("Betrag |Tu|:")
print(np.abs(Tu))
print("Phase φu (in Grad):")
print(np.angle(Tu, deg=True))
```

# Antenna matching Circuit

```python
import numpy as np
pi = np.pi
j = 1j


Rant = 80.
L1 = 2.568e-9
C1 = 642.088e-15
w = 2400e6*2*pi

ZL1 = j*w*L1
ZC1 = 1/(j*w*C1)

ZZ = np.array([[ZL1+ZC1,ZC1],[ZC1, ZC1]])

Z11 = ZZ[0,0]
Z12 = ZZ[0,1]
Z21 = ZZ[1,0]
Z22 = ZZ[1,1]

Ze1 = Z11 - (Z12*Z21)/(Z22 + Rant)
print("Eingangsimpedanz Ze1:")
print(Ze1)
```
