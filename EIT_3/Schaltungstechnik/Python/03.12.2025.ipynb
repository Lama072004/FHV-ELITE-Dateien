# EX01 OPAmp with Feedback

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
Uin = 5.
RL = 20.
Ri = 2e6
Ro = 75.
mu = 200e3

#Feedback
R1 = 3000.
R2 = 1000.
```

```python
RR = np.array([[Ri+R1,R2,0], [R2 - mu*Ri, R2+R1+Ro,Ro], [0-mu*Ri,Ro,Ro+RL]])

UU = np.array([[Uin],[0],[0]])

II = np.linalg.solve(RR,UU)

Vout = RL * II[2]
print("Vout: ", Vout[0], " V")


UOut = Uin*(1+(R1/R2))
print("Vout (ideal): ", UOut, " V")
```

```python
mu_vec = np.logspace(0, 6, 100000)
Vout_vec = []
plt.figure()
for mu in mu_vec:
    RR = np.array([[Ri+R1,R2,0], [R2 - mu*Ri, R2+R1+Ro,Ro], [0-mu*Ri,Ro,Ro+RL]])
    II = np.linalg.solve(RR,UU)
    Vout = RL * II[2]
    Vout_vec.append(Vout[0])
plt.semilogx(mu_vec, Vout_vec)
plt.xlabel('mu')
plt.ylabel('Vout (V)')
plt.grid(True)
plt.show()
```
