```python
import matplotlib.pyplot as plt
import numpy as np

# Parameter (beliebig gewählt)
n_hat = 5
T = 1

# Frequenzindex n
n_max = 10
n_pos = np.arange(1, n_max + 1)
n_complex = np.arange(-n_max, n_max + 1)
n_complex = n_complex[n_complex != 0]  # Remove n=0 since c_0=0

# Reelle Fourierkoeffizienten
b_n = -2 * n_hat / (np.pi * n_pos) * (-1)**n_pos
amplitude_real = np.abs(b_n)

# Komplexe Fourierkoeffizienten
c_n = n_hat / (np.pi * np.abs(n_complex))  # Betrag
amplitude_complex = c_n

# Plots
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 5))

# Reelles Amplitudenspektrum
ax1.stem(n_pos, amplitude_real, basefmt=" ", use_line_collection=True)
ax1.set_xlabel('n')
ax1.set_ylabel('$|b_n|$')
ax1.set_title('Reelles Amplitudenspektrum')
ax1.grid(True, alpha=0.3)
ax1.set_xticks(n_pos)

# Komplexes Amplitudenspektrum
ax2.stem(n_complex, amplitude_complex, basefmt=" ", use_line_collection=True)
ax2.set_xlabel('n')
ax2.set_ylabel('$|c_n|$')
ax2.set_title('Komplexes Amplitudenspektrum')
ax2.grid(True, alpha=0.3)
ax2.set_xticks(n_complex)

plt.tight_layout()
plt.show()
```
